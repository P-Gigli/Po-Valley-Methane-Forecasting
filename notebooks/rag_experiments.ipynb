```python
from po_valley_methane_forecasting.paths import find_project_root
```

```python
from po_valley_methane_forecasting.rag.ingestion import chunk_corpus

from po_valley_methane_forecasting.rag.embedding import (
    load_embedding_model,
    embed_chunks
)

from po_valley_methane_forecasting.rag.retrieval import retrieve_chunks

from po_valley_methane_forecasting.rag.generation import (
    build_context,
    build_generation_prompt,
    generate_answer
)
```

```python
project_root=find_project_root()
```

```python
documents_dir = (
    project_root
    / "rag"
    / "documents"
)

all_chunks = chunk_corpus(
    documents_dir,
    max_chars=1000,
    overlap_sentences=1,
)

print("Documents:", len(
    {chunk["source"] for chunk in all_chunks}
))

print("Total chunks:", len(all_chunks))
```

```python
embedding_model = load_embedding_model()

corpus_embeddings = embed_chunks(
    chunks=all_chunks,
    model=embedding_model,
)
```

```python
print(corpus_embeddings.shape)
print(corpus_embeddings.dtype)
```

```python
query_1 = (
    "Can Sentinel-5P XCH 4 observations be interpreted directly as methane emissions?"
)
```

```python
results_1 = retrieve_chunks(
    query=query_1,
    model=embedding_model,
    corpus_embeddings=corpus_embeddings,
    chunks=all_chunks,
    top_k=5,
    max_per_source=2,
    min_score=0.5,
)
```

```python
results=results_1
print(f"{len(results)} Results!")
for i in range(len(results)):
    print(f"\nResult {i+1}")
    print(f"Score: {results[i]['score']}")
    print(f"Source: {results[i]['source']}")
    print(f"Pages: {results[i]['page_start']}-{results[i]['page_end']}")
    print(results[i]['text'][:500])
```

```python
context_1 = build_context(
    results_1
)
```

```python
prompt_1=build_generation_prompt(query_1, results_1)
```

```python
answer_1, metrics_1 = generate_answer(
    prompt_1
)
```

```python
print(answer_1)
```

```python
print(metrics_1)
```

```python
tokens_per_second = (
    metrics_1["output_tokens"]
    / metrics_1["generation_duration"]
    * 1e9
)

print(f"{tokens_per_second:.2f} tokens/s")
```
